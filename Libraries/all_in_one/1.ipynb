```python
class Dog:

    kind = 'canine'         # class variable shared by all instances

    def __init__(self, name):
        self.name = name    # instance variable unique to each instance

d = Dog('Fido')
e = Dog('Buddy')
print(d.kind)                # shared by all dogs
print(e.kind)                # shared by all dogs
print(d.name)                # unique to d
print(e.name)                # unique to e
```

```python
class Dog:

    tricks = []             # mistaken use of a class variable

    def __init__(self, name):
        self.name = name

    def add_trick(self, trick):
        self.tricks.append(trick)

d = Dog('Fido')
e = Dog('Buddy')
d.add_trick('roll over')
e.add_trick('play dead')
d.tricks                # unexpectedly shared by all dogs
```

```python
class main:
    
    def __init__(self):
        pass

    def func(self):
        return "This is inside the main class"
    

class sub(main):

    def __init__(self):
        pass

    def func(self):  # if we comment this and the next line, auomatically the func() of main class is called for "s.func()"
        return "this is inside the sub class"
    
m = main()
print(f"when the instance of a main class is made and 'func' method is called we get: {m.func()}")
s = sub()
print(f"when the instance of a sub class is made and 'func' method is called we get: {s.func()} \nwe can clearly see the method overriding")
```

```python
class ApnaComplex:

    def __init__(self,a=0,b=0):
        self.a = a
        self.b = b
        
    def __str__(self):
        if self.a == 0:
            return f"({self.b}i)"
        elif self.b<0:
            return f"({self.a}{self.b}i)"
        else:
            return f"({self.a}+{self.b}i)"
        
    def __len__(self):
        return len([self.a,self.b])
    

    def __add__(self,val):
        ans_a = self.a+val.a
        ans_b = self.b+val.b

        ans = ApnaComplex(ans_a,ans_b)
        print(ans)

    def add(self,val):
        return self.__add__(val)

    def __sub__(self,val):
        ans_a = self.a-val.a
        ans_b = self.b-val.b

        ans = ApnaComplex(ans_a,ans_b)
        print(ans)
    def __mul__(self,val):
        ans_a = (self.a*val.a) - (self.b*val.b)
        ans_b = (self.a*val.b) + (self.b*val.a)

        ans = ApnaComplex(ans_a,ans_b)
        print(ans)

    def __getitem__(self,index):
        if index == 0 or index == 1:
            ans = {
                0:self.a,
                1:self.b
            }
            return ans[index]
        else:
            print("index out of scope")


    def __setitem__(self,index,value):
        if index == 0:
            self.a = value
        elif index == 1:
            self.b = value
        else:
            print("index out of scope")

    def __call__(self,message):
        print(f"{self.a} {self.b} {message}")



ap = ApnaComplex(3,4)
ac = ApnaComplex(4,5)
```

```python
print(ap)
print(len(ap))
ap+ac
ap-ac
ap*ac
ap[1]
ap[0] = 100

print(ap)
```
