```python
class Student:

    def __init__(self,name,age):
        self.name = name
        self.age = age

    def __str__(self):
        return (f"This object is of Student class")
    
s1 = Student("Rushit",26)

print(s1)
```

```python
class Student:
    # Class / static variable
    counter = 1

    def __init__(self,name,age):
        self.name = name
        self.age = age 
        
        self.sno = Student.counter 
        Student.counter += 1


c1 = Student("Rushit",23)
c2 = Student("Nitish",34)
c3 = Student("Rakesh",24)

print(c1.sno)
print(c2.sno)
print(c3.sno)
```
