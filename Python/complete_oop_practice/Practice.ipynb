```python
class Parent:
    def __init__(self, name):
        self.name = name  # Parent attribute

    def greet(self):      # Parent method
        print(f"Hello from {self.name}")

class Child(Parent):
    def __init__(self, name, age):
        # 1. Inherit parent attributes
        super().__init__(name) 
        self.age = age    # Child-specific attribute

    def greet(self):
        # 2. Call parent method then add extra logic
        super().greet()   
        print(f"I am {self.age} years old.")

# Usage
obj = Child("Alice", 25)
obj.greet()
```

```python
# If the child class is not having the constructor in that case the parent clas's constructor will be called, and if the child class is having the constructor, in that case the parent class's constructor will not be called, if we want to provide provide values to the parent constructor, in that case we use "super().__init()" syntax and provide the required attributes inside brackets of "init" as shown below

class F:

    def __init__(self, name,age):
        self.name = name
        self.age = age

    def dfg(self):
        return(f"this is in parent class")

class D(F):

    def __init__(self, name,age,sex):
        self.sex = sex
        super().__init__(name,age)

    def dfg(self):
        print(super().dfg())
        return(f"this is in cihld class")

d = D("rushit",26,"male")

d.dfg()
```

```python
# Basically we can not access the child class's attributes in the parent class

class F:

    def __init__(self, name,age):
        self.name = name
        self.age = age
        self.var = 100

    def funcd(self):
        print(f"{self.name} is in parent class") # This is fine but
        # print(f"{self.sex} is in parent class") # This will throw an error 

class D(F):

    def __init__(self,name,age,sex):
        super().__init__(name,age)
        self.sex = sex

    def funcd(self):
        print(f"{self.sex} is in child class")  # this will obviosuly work
        print(f"{self.var} is in parent class, but called inside the child class")  # even this is possible becasue we have called the parent's constructor
        super().funcd()

    
d =D("Ruhsit",26,"male")

d.funcd()
```
