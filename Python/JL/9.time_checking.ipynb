```python
import timeit
```

```python
def sqr():
    a = [1,2,3,4,5,6,7,8,9,1,2,3,4,5,6,7,8,9,1,2,3,4,5,6,7,8,9,1,2,3,4,5,6,7,8,9,1,2,3,4,5,6,7,8,9]
    b = []
    for i in a:
        b.append(i*i)
    return b

for_execution_time = timeit.timeit(stmt=sqr,number=100)

def lc():
    a = [1,2,3,4,5,6,7,8,9,1,2,3,4,5,6,7,8,9,1,2,3,4,5,6,7,8,9,1,2,3,4,5,6,7,8,9,1,2,3,4,5,6,7,8,9]
    return [i*i for i in a]

lc_execution_time = timeit.timeit(stmt=lc,number=100)

def sqr(i):
    return i*i
def mp():
    a = [1,2,3,4,5,6,7,8,9,1,2,3,4,5,6,7,8,9,1,2,3,4,5,6,7,8,9,1,2,3,4,5,6,7,8,9,1,2,3,4,5,6,7,8,9]
    return(sqr,a)

mp_execution_time = timeit.timeit(stmt=mp,number=100)

print(f"for_execution_time = {for_execution_time}")
print(f"lc_execution_time = {lc_execution_time}")
print(f"mp_execution_time = {mp_execution_time}")
```
