- Identity Operator

"is" and "is not"

```python
# this operator will compare the memory address of both the variables and returm answer accordingly

a = 5
b = 5
c = "5"   # because the data is not same it is not stored in the same memory addres
d = "5"

print(id(a))
print(id(b))
print(id(c))

print(a is b)
print(a is c)

print(a is not b)
print(a is not c)
print(c is d)
```

```python
# nested if-else
# if-else block inside another if-else block "Indentation is very important"

number  = int(input("enter the number"))

if number%2 == 0:
    if number <= 100:
        print(f"{number} is even and less or equal to hundred")
    else:
        print(f"{number} is even but greater than hundred")
else:
    if number <= 100:
        print(f"{number} is even and less or equal to hundred")
    else:
        print(f"{number} is even but greater than hundred")
```

```python
year = int(input("enter the year"))

if year%4 == 0:
    if year%100==0:
        if year%400==0:
            print(f"{year} is a leap year")
        else:
            print(f"{year} is not a leap year")   
    else:
        print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
```
