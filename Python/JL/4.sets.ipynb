```python
a = {1,2,3,4,5,6,7}
print(a)
a.add(100)  # add element
print(a)
a.remove(4)  # remove element, but throws error if any element is not present in set
print(a)
a.discard(3)
print(a) # if the elemet is not their in the set, it won't throw any error
print(a.pop()) # Gives the random item from the set
print(a) # BTW the popped item is removed from the set
# a.delete(2) --> This will throw an error
```

```python
a = {1,2,3,4,5,6}
b = {4,5,6,7,8,9}
c = {7,8,9,10,11,12}

# union
print(a.union(b))  # all the elements from both the sets
print(b.union(a,c))
print(a | b)  # another method, al of them should be sets only
print(b | a | c)  # also we can use multiple sets
a.update(b)  # all the elements will be added to first set, makes changes inplace and returns nothing
print(a)
```

```python
a = {1,2,3,4,5,6}
b = {4,5,6,7,8,9}
c = {4,5,6,7,8,9,10,11,12}
# intersection
print(a.intersection(b))
print(b.intersection(b,c))
print(a&b)
print(a&b&c)
a.intersection([1,2,3]) # only the common elemets from both the sets will be added to first set, makes change inplace and returns nothing
print(a)
```

```python
a = {1,2,3,4,5,6}
b = {4,5,6,7,8,9}
c = {4,5,6,7,8,9,10,11,12}
# difference
print(a.difference(b))  # returns all the elements from set1, but by removing the common elements in both the sets
print(a-b)  # same thing as above
print(a.difference(b,c))
print(a-b-c) # hope you understand
a.difference_update(b) # same as difference, but makes changes in the first set inplace itself
print(a)
```

```python
a = {1,2,3,4,5,6}
b = {4,5,6,7,8,9}
c = {4,5,6,7,8,9,10,11,12}

print(a.symmetric_difference(b)) # returns all the elements except the common elements, basucally (a union b) - (a intrseciton b)
print(a^b) # same as above
print(a^b^c) 
```

```python
a = {1,2,3,4,5,6}
b = {4,5,6,7,8,9}
c = {4,5,6,7,8,9,10,11,12}
d = {1,2,3}

print(a.isdisjoint(b)) # baically true if they both have no common element
print(b.isdisjoint(d)) # True becasue no common elements
print(d.issubset(a)) # basically true if all the elements of 1st set are present in 2nd set and vice versa
print(d <= a) # another way of checking subset
print(a.issuperset(d)) # basically reverse of subset
print(d >= a)
```
