- collections in python is nothing but the container data structures thar are used for storing and organizing data, such as lists, tuples, dictionaries, and sets
- The "collections" module in python contains the specialized data type, they are Counter, deque, defaultdict, OrderedDict, ChainMap, namedtuple(), UserString

```python
# namedtuple

from collections import namedtuple

a = namedtuple("course",'name, technology')

s = a('python','data science')
print(s)
```

```python
a = namedtuple("course",'name, technology')

s = a._make(['python','data science'])
print(s)
```

```python
from collections import deque # the optimized way of insertion and deletion in the lisr

a = [1,2,3,4,5,6,7,8,9]
d = deque(a)

print(d)
d.append(10)
print(d)
d.appendleft(0)
print(d)
d.pop()
print(d)
d.popleft()
print(d)
```

```python
from collections import ChainMap

a = {'name':"Rushit","surname":"Patel"}
b = {"course":"Python","technology":"Ds"}

cm = ChainMap(a,b)
cm
```

```python
from collections import Counter

a = [1,2,3,4,5,6,2,3,4,5,6,4,5,6,3,4,5,6,1,2,3,2,5,2,1]

c = Counter(a)
print(c)
print(list(c.elements()))
print(c.most_common())
sub = {2:1,5:2}
c.subtract(sub)
print(c.most_common())
```

```python
from collections import OrderedDict

# it is nothing but the dictionary subclass which remembers the order in which the entries were done

d = OrderedDict()

d[1] = 'r'
d[2] = 'u'
d[3] = 's'
d[4] = "h"
d[5] = 'i'
d[6] = 't'

print(d)

d[1] = 'R'
d[2] = 'U'
d[3] = 'S'

print(d)
```

```python
from collections import defaultdict

# Basically it will not throw any error even if the key which is called is not existing in the dictionary

d = defaultdict(int)

d['nmae'] = "Rushit"
d["sur_name"] = "Patel"

print(d['age'])
```

```python
li = ['apple', 'banana', 'apple', 'orange', 'banana', 'apple']

c = Counter(li)
c
```

```python
print(li)

gl = defaultdict(list)

for i in li:
    key = len(i)
    gl[key].append(i)
gl
```
