- tuples are immutable (we can modify the values once a tuple is created)
- this are ordererd data structures
- this can have duplicate values
- they can have multiple data types (Heterogeneous)

```python
# Immutable
a = (1,2,3,4,5,6,7,8,9)
#we can not add, remove or modify any elements in the tuple

# a.remove(3), a.pop(3), a.delete(3)  --> all this will throw an error

# a.append(3), a.add(3), a.extend(3) --> all this will also throw error
```

```python
a = (1,2,3,4,5,6,7,8,9)
print(a)
a = (1,"Rushit",[1,2,3],(1,2),4.5,True,{"message":"This can also be in a list"}) # this is a valid tuple
print(a)
```

```python
a = (1,2,3,4)
print(type(a))
a = (1,)  # this is still a tuple
print(type(a))
a = (1)   # this will be int not tuple
print(type(a))
```

```python
a = (1,2,3,4.9,"ram",[1,2,3,4],(3,4,5,6),True)

print(a[0])
print(a[-1])
print(a[0:])
print(a[6][0])
print(a[6])
print(len(a))
```

```python
a = (1,2,3,3,3,3,3,3)
b = (4,5,6.0)
print(id(a))
a = a+b
print(id(a))   # see the address is changed

print(min(a))
print(max(a)) # works only with integers, float, not with mixed datatype tuples
print(a.count(3)) # gives the count of the element inn the tuple

a = [1,2,3,4,5,6]
print(a,type(a))
a = tuple(a) # convert the list into tuple
print(a,type(a))

a = (1,)
print(a)
a = a*5
print(a)
```
