# sort it up 
- given an array "arr" of length n, consists of only 0's, 1's and 2's, our goal is to sort the array in assecinding order without using sort() function

```python
def sort_array(arr):
    l = 0
    c = 0
    r=len(arr)-1

    while c <= r:
        if arr[c] == 0:
            arr[l],arr[c] = arr[c],arr[l]
            l += 1
            c += 1

        elif arr[c] == 1:
            c += 1

        else:
            arr[c],arr[r] =arr[r],arr[c]
            r-=1
    return arr
```

```python
a = [2,1,0,1,2]
sort_array(a)
```

```python
a2 = [1,2,1,0,1,1,0,0,0,2,0,2,2,2,1,1,0]
sort_array(a2)
```

1. merging 3 arrays
- We have thre sorted arrays, just pick the common elements in this three arrays and give the output

```python
def merge_common(a1,a2,a3):
    res=[]
    for i in a1:
        if i in a2:
            if i in a3:
                res.append(i)
    return res
```

```python
a1 = [1,2,3,4,5]
a2 = [1,3,4,5,8]
a3 = [1,2,5,7,9]

merge_common(a1,a2,a3)


t1 = time.time()
time.sleep(2)
merge_common(a1,a2,a3)
print(time.time()-t1)
```

```python
# Actual solution

def merge_common_new(a1,a2,a3):
    
    x_loc = 0
    y_loc = 0
    z_loc = 0
    res=[]

    while x_loc != len(a1) and y_loc != len(a2) and z_loc != len(a3):

        x = a1[x_loc]
        y = a2[y_loc]
        z = a3[z_loc]

        if x == y and y == z:
            res.append(x)
            x_loc+=1
            y_loc+=1
            z_loc+=1

        elif x<=y and x<=z:
            x_loc+=1
        elif y<=x and y<=z:
            y_loc+=1
        else:
            z_loc+=1
    return res
```

```python
a1 = [1,2,3,4,5]
a2 = [1,3,4,5,8]
a3 = [1,2,5,7,9]

t1 = time.time()
time.sleep(2)
merge_common_new(a1,a2,a3)
print(time.time()-t1)
```

#  Missing no.
- given 'n' an integer and "arr" which has values from 1 to n but length of array is n-1 basically no duplicates and one value is missing, we have to find that missing value

```python
a1 = [1,2,4,5,6,7,8,9]
n=9
t1 = time.time()
time.sleep(2)
print( ((n*(n+1))/2) - sum(a1) )            # we know that sum of n number starting from 1 to n is ( n*(n-1) )/2  so if we substract the sum(values in array) we get the result
print(time.time()-t1)
```

```python
# Actual solution

def missingno(n,a1):
    total = (n*(n+1))/2
    a_sum=0
    
    for i in range(n-1):
        a_sum = a_sum + a1[i]
    return total-a_sum
```

```python
a1 = [1,2,4,5,6,7,8,3]
n=9

t1 = time.time()
time.sleep(2)
missingno(9,a1)
print(time.time()-t1)
```

```python
# Actual solution

def sort_squared_array(a1):
    r = []
    p = []
    n = []

    for i in a1:
        if i<0:
            n.append(i)
        else:
            p.append(i)
    p_idx = 0
    n_idx = len(n)-1

    while p_idx < len(p) and n_idx >=0:
        if p[p_idx]**2 < n[n_idx]**2:
            r.append(p[p_idx]**2)
            p_idx+=1
        else:
            r.append(n[n_idx]**2)
            n_idx-=1
    while p_idx < len(p):
        r.append(p[p_idx]**2)
        p_idx+=1
    while n_idx >=0:
        r.append(n[n_idx]**2)
        n_idx-=1
    return r
```
